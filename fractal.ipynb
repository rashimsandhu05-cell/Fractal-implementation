```python
import turtle


def draw_triangle(points, colour):
    """Draw a filled triangle using three points."""
    
    turtle.fillcolor(colour)
    turtle.penup()
    turtle.goto(points[0])
    turtle.pendown()

    turtle.begin_fill()

    for point in points[1:]:
        turtle.goto(point)

    turtle.goto(points[0])
    turtle.end_fill()


def midpoint(point1, point2):
    """Return the midpoint between two points."""

    return (
        (point1[0] + point2[0]) / 2,
        (point1[1] + point2[1]) / 2
    )


def sierpinski(points, degree):
    """Recursively draw a Sierpinski Triangle."""

    colours = [
        "blue",
        "red",
        "green",
        "yellow",
        "purple",
        "orange"
    ]

    draw_triangle(points, colours[degree % len(colours)])

    if degree > 0:

        # Top triangle
        sierpinski(
            [
                points[0],
                midpoint(points[0], points[1]),
                midpoint(points[0], points[2])
            ],
            degree - 1
        )

        # Bottom-left triangle
        sierpinski(
            [
                points[1],
                midpoint(points[0], points[1]),
                midpoint(points[1], points[2])
            ],
            degree - 1
        )

        # Bottom-right triangle
        sierpinski(
            [
                points[2],
                midpoint(points[0], points[2]),
                midpoint(points[1], points[2])
            ],
            degree - 1
        )


# Set up the turtle window
screen = turtle.Screen()
screen.title("Sierpinski Triangle Fractal")
screen.bgcolor("white")

# Make drawing faster
turtle.speed(0)
turtle.hideturtle()

# The three corners of the main triangle
points = [
    (-300, -250),
    (0, 300),
    (300, -250)
]

# Recursion depth
degree = 5

# Draw the fractal
sierpinski(points, degree)

# Keep the window open
turtle.done()
```
